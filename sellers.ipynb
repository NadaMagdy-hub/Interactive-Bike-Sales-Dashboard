----تنظيف أسماء المدن (Lowercase و شيل المسافات)-------

```python
sellers['seller_city'] = sellers['seller_city'].astype(str).str.strip().str.lower()
```

---- تحويل الولايات لحروف كبيرة (Uppercase) عشان تطابق ملف الجيولكيشن بالظبط (SP, RJ...)-----

```python
sellers['seller_state'] = sellers['seller_state'].astype(str).str.strip().str.upper()
```

-----نسيب الـ Zip Code كـ Integer (رقم) زي ملف الجيولكيشن عشان يربطوا صح------

```python
sellers['seller_zip_code_prefix'] = sellers['seller_zip_code_prefix'].astype(int)
```
